```python

import paho.mqtt.client as mqtt


# Configuración MQTT
broker = "broker.hivemq.com"
puerto = 1883
topic = "carlos"

# Convertir binario a texto
def convertir_binario_a_texto(binario):
    caracteres = binario.split()

    texto = "".join(chr(int(byte, 2)) for byte in caracteres)

    return texto


# Cuando se conecta al broker
def on_connect(cliente, userdata, flags, rc):
    if rc == 0:
        print("Conectado al broker MQTT")
        print("Suscrito al topic:", topic)
        cliente.subscribe(topic)
    else:
        print("Error de conexión. Código:", rc)


# Cuando recibe un mensaje
def on_message(cliente, userdata, mensaje):
    binario_recibido = mensaje.payload.decode()

    print("\nBinario recibido:")
    print(binario_recibido)

    try:
        texto = convertir_binario_a_texto(binario_recibido)

        print("\nTexto convertido:")
        print(texto)

    except Exception as error:
        print("Error al convertir el mensaje:", error)


# Crear cliente MQTT
cliente = mqtt.Client()

# Asignar funciones
cliente.on_connect = on_connect
cliente.on_message = on_message

# Conectar al broker
print("Conectando al broker...")

cliente.connect(broker, puerto, 60)

# Mantener conexión y recibir mensajes
cliente.loop_forever()
```
